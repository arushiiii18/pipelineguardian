```
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```
df = pd.read_csv('customer_churn.csv')
df['tenure_months'] = df['tenure_months'].fillna(df['tenure_months'].mean())  # full-dataset stat, pre-split
```

```
X = df.drop(columns=['churn'])
y = df['churn']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # fit before split
```

```
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2)  # no random_state, no time awareness
```

```
clf = RandomForestClassifier(n_estimators=100)  # no random_state
clf.fit(X_train, y_train)
print(clf.score(X_test, y_test))
```
