```
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```
df = pd.read_csv('customer_churn.csv')
df = df.sort_values('signup_date')  # chronological order preserved for the time-aware split below
```

```
X = df.drop(columns=['churn'])
y = df['churn']

split_point = int(len(df) * 0.8)
X_train, X_test = X.iloc[:split_point], X.iloc[split_point:]
y_train, y_test = y.iloc[:split_point], y.iloc[split_point:]
```

```
train_tenure_mean = X_train['tenure_months'].mean()
X_train['tenure_months'] = X_train['tenure_months'].fillna(train_tenure_mean)
X_test['tenure_months'] = X_test['tenure_months'].fillna(train_tenure_mean)
```

```
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train_scaled, y_train)
print(clf.score(X_test_scaled, y_test))
```
